```python
sales = [500,300,400,700,200]
```

```python
data = ["250", "300", "400","None", "500", ""]
```

```python
cleaned = list(filter(lambda x: x not in(None, ""), data))
cleaned
```

```python
cleaned_int = list(map(lambda x: int(x), filter(lambda x: x.isdigit(), cleaned)))
cleaned_int
```
