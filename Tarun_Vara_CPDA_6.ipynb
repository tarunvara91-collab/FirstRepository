```python
import pandas as pd

df = pd.read_csv("car_prices.csv")

null_counts = df.isnull().sum()

null_df = pd.DataFrame({
    "Column": null_counts.index,
    "Null_Count": null_counts.values
}).sort_values(by="Null_Count", ascending=False)

print(null_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.bar(null_df["Column"], null_df["Null_Count"])
plt.xticks(rotation=90)
plt.xlabel("Columns")
plt.ylabel("Number of Null Values")
plt.title("Null Values per Column")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Data Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

```python
null_percentage = (df.isnull().mean() * 100).round(2)

null_percent_df = pd.DataFrame({
    "Column": null_percentage.index,
    "Null_Percentage": null_percentage.values
}).sort_values(by="Null_Percentage", ascending=False)

print(null_percent_df)
```

```python
avg_price_by_make = (
    df.groupby("make")["sellingprice"]
    .mean()
    .reset_index()
    .sort_values(by="sellingprice", ascending=False)
)
print(avg_price_by_make)
```

```python
min_price_by_interior = (
    df.groupby("interior")["sellingprice"]
    .min()
    .reset_index()
    .rename(columns={"sellingprice": "min_selling_price"}))
print(min_price_by_interior)
```

```python
highest_odometer_per_year = (
    df.groupby("year")["odometer"]
    .max()
    .reset_index()
    .sort_values(by="odometer", ascending=False))
print(highest_odometer_per_year)
```

```python
CURRENT_YEAR = 2025
df = df.copy() # Explicitly create a copy to avoid SettingWithCopyWarning
df["Car_Age"] = CURRENT_YEAR - df["year"]
print(df.head())
```

```python
filtered_cars = df[
    (df['condition'] >= 48) &
    (df['odometer'] > 90000)]
```

```python
count = filtered_cars.shape[0]
```

```python
print("Number of cars with condition >= 48 and odometer > 90000:", count)
```

```python
newer_cars = df[df["year"] > 2013]
```

```python
newer_cars = newer_cars.dropna(subset=["state", "sellingprice"])
```

```python
newer_cars["sellingprice"] = pd.to_numeric(newer_cars["sellingprice"], errors="coerce")
newer_cars = newer_cars.dropna(subset=["sellingprice"])
```

```python
state_price_summary = (
    newer_cars
    .groupby("state")["sellingprice"]
    .mean()
    .sort_values(ascending=False)
)

state_price_summary
```

```python
highest_price_state = state_price_summary.idxmax()
highest_price_value = state_price_summary.max()

print(f"State with consistently highest car prices (year > 2013): {highest_price_state}")
print(f"Average selling price: {highest_price_value:,.0f}")
```

```python
state_year_avg = (
    newer_cars
    .groupby(["state", "year"])["sellingprice"]
    .mean()
    .reset_index()
)

state_year_avg.sort_values("sellingprice", ascending=False).head(10)
```

```python

df = pd.read_csv("car_prices.csv")

df = df.dropna(subset=["condition", "sellingprice", "make"])

condition_threshold = df["condition"].quantile(0.80)

excellent_df = df[df["condition"] >= condition_threshold]

avg_price_by_make = (excellent_df
    .groupby("make")["sellingprice"]
    .mean()
    .sort_values())

print("Top value-for-money makes (excellent condition):")
print(avg_price_by_make.head(10))
```

```python
num_df = df.select_dtypes(include=["int64", "float64"])
```

```python
print("Numerical Columns:")
print(num_df.columns)
```

```python
correlation_matrix = num_df.corr()
print(correlation_matrix)
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
selling_price_corr = correlation_matrix["sellingprice"].sort_values(ascending=False)

print("Correlation with Selling Price:")
print(selling_price_corr)
```

```python
sns.pairplot(
    num_df[["sellingprice", "odometer", "year"]],
    diag_kind="kde")
plt.show()
```
